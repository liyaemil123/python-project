# **Student Performance Analysis**




# **OBJECTIVE**

The objective of this projects is to analyze student performance and study habits using Python, Pandas, and NumPy. The project aims to identify relationships between factors such as study hours, attendance, assignment completion, and academic marks. By analyzing the data, the project seeks to identify patterns that may influence students' academic performance and provide meaningful insights into the factors associated with better learning outcomes.

# **Problem Definition**

Student academic performance can be influenced by several factors such as study hours, attendance, assignment completion, sleep patterns, and previous academic performance. However, it can be difficult to understand which factors have the greatest impact on students' final marks when looking at individual records.

This project aims to analyze student performance data using Python, Pandas, and NumPy to identify patterns and relationships between students' study habits, attendance, assignment completion, sleep hours, previous semester marks, and final marks. By analyzing and comparing these factors, the project seeks to understand the factors associated with better academic performance and identify areas that may help students improve their learning outcomes.

**1. **DATA** LOADING**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("student_performance.csv")

df
```

**3. Find Highest and Lowest Marks**

```python
df["Final_Marks"].max()
```

```python
df["Final_Marks"].min()
```

**4. Select Specific Columns**

```python
# Select specific columns
df["Study_Hours_Per_Day"]
```

```python
#Select Multiple Columns
df[["Student_ID", "Study_Hours_Per_Day", "Final_Marks"]]
```

**5. Filter Students**

```python
# Students with final marks above 80
high_performers = df[df["Final_Marks"] > 80]

high_performers
```

```python
# Students studying more than 6 hours

df[df["Study_Hours_Per_Day"] > 6]
```

```python
# Students with attendance above 90%
df[df["Attendance_Percentage"] > 90]
```

**6. Grade Distribution**

```python
df["Grade"].value_counts()
```

```python
#Sort Grades
df["Grade"].value_counts().sort_index()
```

**7. Use NumPy**

```python
marks = df["Final_Marks"].to_numpy()

print(marks)
```

```python
print("Mean:", np.mean(marks))
print("Maximum:", np.max(marks))
print("Minimum:", np.min(marks))
print("Standard Deviation:", np.std(marks))
```

**8. Find Top Performing Students**

```python
top_students = df.sort_values(
    by="Final_Marks",
    ascending=False
).head(10)

top_students
```

**9. Visualization**

```python

#Grade distribution

df["Grade"].value_counts().plot(kind="bar")

plt.title("Grade Distribution")
plt.xlabel("Grade")
plt.ylabel("Number of Students")
plt.show()
```

```python
#Study hours vs final marks

plt.scatter(
    df["Study_Hours_Per_Day"],
    df["Final_Marks"]
)

plt.title("Study Hours vs Final Marks")
plt.xlabel("Study Hours Per Day")
plt.ylabel("Final Marks")
plt.show()
```
